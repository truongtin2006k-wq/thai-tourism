### 1.4 Collection method

**Tầng 1: số liệu gốc của MOTS.** Số liệu được Bộ Du lịch và Thể thao Thái Lan (MOTS) công bố
dưới dạng các workbook Excel thống kê du lịch nội địa cấp tỉnh trên mots.go.th. Mô tả của tác giả
Kaggle không nói MOTS thu thập hoặc ước tính các chỉ số bằng phương pháp nào (khảo sát, báo cáo
hành chính...), nên phần này **chưa được xác minh**; đây là một hạn chế khi diễn giải kết quả.

**Tầng 2: cách tác giả Kaggle dựng bảng dữ liệu** (theo mô tả của tác giả):
1. Tải 7 workbook theo tháng cấp tỉnh được liệt kê trong `build_report.csv` từ mots.go.th.
2. Đọc từng sheet tháng, bỏ các sheet cộng dồn từ đầu năm và sheet năm tài chính. Tháng được lấy
   từ **tên sheet**, vì có 4 sheet ghi sai tháng hoặc năm ở tiêu đề; việc căn chỉnh tháng được xác
   nhận bằng cách so cột cùng kỳ năm trước với các workbook khác.
3. Chuẩn hóa tên tỉnh: bỏ ký hiệu chú thích và sửa các cách viết sai chính tả thành tên chính thức.
   Mỗi nhãn dòng phải khớp một trong 77 tỉnh, một dòng tổng vùng, tổng toàn quốc hoặc ghi chú,
   nếu không quá trình dựng dừng lại.
4. Với mỗi bộ (tỉnh, tháng, chỉ số), giữ giá trị từ workbook được công bố gần nhất.
5. Tác giả báo cáo các kiểm tra đều đạt, gồm: tổng 77 tỉnh bằng tổng toàn quốc đã công bố ở 7 chỉ
   số cộng gộp được, và số khách Thái cộng khách nước ngoài bằng tổng khách ở mọi dòng.
6. Không nội suy giá trị nào. 49 ô trong workbook 2021 ghi "-" (nghĩa là không có) và được đọc là 0;
   workbook 2020 cũng ghi 0 ở đúng các ô đó. 3 ô trống của workbook 2020 (tháng 4/2020) **không được
   dùng**, vì giá trị cho 2020 lấy từ workbook 2021 và ở đó là 0.

**Những điều tôi đã tự kiểm tra lại trên bảng chính:** 77 tỉnh; 93 tháng liên tục từ 01/2019 đến
09/2026; 8 chỉ số; không có khóa (date, province_en, measure) nào bị lặp; số dòng thực tế bằng số
dòng kỳ vọng (57.288 = 77 × 93 × 8); không có giá trị thiếu (NaN).

**Quần thể và khung thời gian:** toàn bộ 77 tỉnh/thành Thái Lan, theo dõi theo tháng từ 01/2019
đến 09/2026 (93 tháng), 8 chỉ số về số khách, doanh thu và tỷ lệ lấp đầy cơ sở lưu trú.

**Hạn chế và sai lệch đã biết:**
- **Số liệu sơ bộ không chỉ ở các tháng gần nhất.** Theo cột `status`, toàn bộ năm 2020 và 2021
  là số sơ bộ (`preliminary`); 2019 và 2022-2025 là số đã hiệu chỉnh (`revised`); năm 2026 có 1.848
  dòng `revised` (tương ứng 3 tháng) và 3.696 dòng `preliminary` (tương ứng 6 tháng). Khi so sánh
  giữa các năm, cần nhớ rằng mức độ chính thức của số liệu không đồng đều.
- **Các chỉ số không đồng nhất về định nghĩa.** Khách "người thăm" (`visitors_*`) gồm cả khách đi về
  trong ngày, còn `guests` chỉ đếm khách lưu trú; vì vậy hai nhóm này không so sánh trực tiếp
  được. `visitors_foreign` là lượt thăm theo từng tỉnh, không phải số khách nhập cảnh cả nước, nên
  tổng 77 tỉnh có thể đếm trùng một khách thăm nhiều tỉnh. `occupancy_rate_pct` không cộng gộp được.
- **Số 0.** Một số số 0 xuất phát từ ô "-" (không có) trong workbook gốc; cần phân tích riêng ở mục 2.4.
- **Chất lượng nguồn gốc.** Workbook có lỗi định dạng và tên tỉnh viết sai; tác giả Kaggle đã xử
  lý như mô tả ở trên, nhưng tôi chưa tự đối chiếu lại với workbook gốc.

### 2.2 Numerical



Dữ liệu dạng long có cột `value` trộn nhiều đơn vị (người, triệu baht, %), nên không thể thống kê trực tiếp trên `value`. Tôi pivot sang dạng wide (mỗi chỉ số một cột, mỗi dòng một cặp tỉnh–tháng) để tính từng chỉ số và so sánh các chỉ số trên cùng một dòng.

##### 2.2.0 Chuyển sang dạng wide

```python
# [CODE] 2.2.0
import numpy as np
import matplotlib.pyplot as plt

wide = df.pivot_table(
    index=["date", "province_en", "region", "status"],
    columns="measure", values="value", aggfunc="first",
    observed=True,                      # tránh tạo tổ hợp rỗng khi cột là category
).reset_index()
wide.columns.name = None
wide["date"] = pd.to_datetime(wide["date"])

count_cols = ["guests", "visitors_total", "visitors_thai", "visitors_foreign"]
rev_cols   = ["revenue_total_mthb", "revenue_thai_mthb", "revenue_foreign_mthb"]
num_cols   = count_cols + rev_cols + ["occupancy_rate_pct"]

print("Số dòng wide:", wide.shape[0], "| kỳ vọng:", df["province_en"].nunique() * df["date"].nunique())
print(wide[num_cols].isna().sum().sum(), "ô NaN")
wide
```

Wide có `<7161>` dòng, khớp `<77 × 93>`.

##### 2.2.1 Thống kê mô tả

```python
# [CODE] 2.2.1
summary = wide[num_cols].describe(percentiles=[.01, .25, .5, .75, .99]).T
summary["IQR"]      = wide[num_cols].quantile(.75) - wide[num_cols].quantile(.25)
summary["skew"]     = wide[num_cols].skew()
summary["pct_zero"] = (wide[num_cols] == 0).mean() * 100
summary.round(2)
```

**Nhận xét:** `<mean so với median của các cột chính; skew; khoảng giữa phân vị 99% và max; cột nào có pct_zero cao (xử lý ở 2.4)>`. Vì các cột bị lệch phải, tôi dùng **median và IQR** để mô tả tâm và độ phân tán.

##### 2.2.2 Hình dạng phân phối

Số khách và doanh thu lệch phải rất mạnh (vài tỉnh lớn kéo đuôi), nên vẽ cả thang thường và thang log10. Dòng bằng 0 bị loại khỏi biểu đồ log vì log(0) không xác định.

```python
# [CODE] 2.2.2
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for i, col in enumerate(["visitors_total", "revenue_total_mthb"]):
    s = wide[col][wide[col] > 0]
    print(col, "- loại", (wide[col] == 0).sum(), "dòng bằng 0 khỏi biểu đồ log")
    axes[i, 0].hist(s, bins=50)
    axes[i, 0].set_title(f"{col} (thang thường)")
    axes[i, 0].set_ylabel("Số dòng (tỉnh-tháng)")
    axes[i, 1].hist(np.log10(s), bins=50)
    axes[i, 1].set_title(f"{col} (log10)")
    axes[i, 1].set_xlabel("log10(giá trị)")
plt.tight_layout(); plt.show()

wide["occupancy_rate_pct"].hist(bins=50)
plt.title("Tỷ lệ lấp đầy cơ sở lưu trú"); plt.xlabel("%"); plt.ylabel("Số dòng"); plt.show()
```

**Nhận xét:** `<hình dạng thang thường; thang log có đối xứng hơn không, có hai đỉnh không; occupancy dồn ở đâu>`.

##### 2.2.3 Min/max và giá trị bất khả thi

```python
# [CODE] 2.2.3
print("Giá trị âm:\n", (wide[num_cols] < 0).sum())
print(wide["occupancy_rate_pct"].agg(["min", "max"]), "| >100:", (wide["occupancy_rate_pct"] > 100).sum())
placeholders = [-1, -999, 999, 9999, 99999]
print("Placeholder:\n", wide[num_cols].isin(placeholders).sum())
print("Số người không nguyên:\n", (wide[count_cols] % 1 != 0).sum())

print(wide.loc[wide["visitors_total"].idxmax(), ["province_en", "date", "visitors_total"]])
```

**Nhận xét:** `<số giá trị âm, occupancy > 100, placeholder, số lẻ>`. Điểm max `visitors_total` thuộc `<tỉnh, tháng>`, `<hợp lý / bất thường>` vì `<lý do>`.

##### 2.2.4 Ràng buộc nội tại giữa các cột

Các cột phải khớp nhau, nên vi phạm là dấu hiệu lỗi. Chọn dung sai sau khi nhìn phân phối chênh lệch, không đoán trước.

```python
# [CODE] 2.2.4a: nhìn chênh lệch trước
gap_v = wide["visitors_total"] - wide["visitors_thai"] - wide["visitors_foreign"]
gap_r = wide["revenue_total_mthb"] - wide["revenue_thai_mthb"] - wide["revenue_foreign_mthb"]
print(gap_v.abs().describe()); print(gap_r.abs().describe())
```

```python
# [CODE] 2.2.4b: đếm vi phạm theo dung sai đã chọn (chỉnh theo kết quả trên)
TOL_V, TOL_R = 1, 0.05
bad = (gap_v.abs() > TOL_V) | (gap_r.abs() > TOL_R)
print(bad.sum(), "dòng vi phạm")
print(wide[bad].groupby([wide["date"].dt.year, "status"]).size())
print("guests > visitors_total:", (wide["guests"] > wide["visitors_total"]).sum())
```

**Nhận xét:** dung sai `<...>` được chọn vì `<lý do dựa trên phân phối chênh lệch>`. `<N>` dòng vi phạm, tập trung ở `<năm/status>`. `guests` và `visitors` có định nghĩa khác nhau (mục 1.4), nên `<kết luận về guests > visitors_total>`.

##### 2.2.5 Outlier

IQR toàn cục sẽ đánh dấu các tỉnh lớn (Bangkok, Phuket…) mọi tháng, vì khác biệt quy mô. Em thử IQR toàn cục rồi chuyển sang robust z-score **trong từng tỉnh**.

```python
# [CODE] 2.2.5a: IQR toàn cục
col = "visitors_total"
q1, q3 = wide[col].quantile([.25, .75]); iqr = q3 - q1
mask = (wide[col] < q1 - 1.5 * iqr) | (wide[col] > q3 + 1.5 * iqr)
print(mask.sum(), "dòng,", wide.loc[mask, "province_en"].nunique(), "tỉnh")
print(wide.loc[mask, "province_en"].value_counts().head(10))
```

```python
# [CODE] 2.2.5b: robust z-score trong từng tỉnh (trên log)
cols = ["visitors_total", "revenue_total_mthb", "guests"]
lg  = np.log1p(wide[cols])
med = lg.groupby(wide["province_en"]).transform("median")
mad = (lg - med).abs().groupby(wide["province_en"]).transform("median")
rz  = 0.6745 * (lg - med) / mad.replace(0, np.nan)
out = rz.abs() > 3.5
print(out.sum())

out["visitors_total"].groupby(wide["date"]).sum().plot(figsize=(11, 3.5),
    title="Số tỉnh có visitors_total bất thường theo tháng")
plt.ylabel("Số tỉnh"); plt.xlabel("Tháng"); plt.show()
```

```python
# [CODE] 2.2.5c: tìm điểm đột biến cô lập (nghi lỗi)
w = wide.sort_values(["province_en", "date"]).copy()
roll = w.groupby("province_en")["visitors_total"].transform(
    lambda s: s.rolling(5, center=True, min_periods=3).median())
w["ratio"] = w["visitors_total"] / roll
sus = w[(w["ratio"] > 5) | (w["ratio"] < 0.1)]
print(len(sus))
sus[["province_en", "date", "visitors_total", "ratio"]].head(20)
```

**Nhận xét:** IQR toàn cục gắn `<X>` dòng thuộc `<Y>` tỉnh, `<chủ yếu là tỉnh lớn>`, nên phản ánh quy mô chứ không phải lỗi. Robust z-score cho thấy outlier dồn vào `<giai đoạn>`, `<khớp / không khớp>` cú sốc COVID. Điểm đột biến cô lập: `<N>` dòng, `<nghi lỗi hay thật>` vì `<lý do>`. Quyết định: `<giữ / xử lý>`.

##### 2.4.3 Phân loại 0 thật và thiếu

```python
n = len(wide)
checks = {
    "mọi 8 chỉ số đều 0": (wide[num_cols] == 0).all(axis=1),
    "visitors_total=0 nhưng revenue_total>0": (wide["visitors_total"] == 0) & (wide["revenue_total_mthb"] > 0),
    "visitors_total>0 nhưng revenue_total=0": (wide["visitors_total"] > 0) & (wide["revenue_total_mthb"] == 0),
    "guests=0 nhưng visitors_total>0": (wide["guests"] == 0) & (wide["visitors_total"] > 0),
    "occupancy=0 nhưng guests>0": (wide["occupancy_rate_pct"] == 0) & (wide["guests"] > 0),
    "guests=0 nhưng occupancy>0": (wide["guests"] == 0) & (wide["occupancy_rate_pct"] > 0),
    "ngoại: visitors=0 nhưng revenue>0": (wide["visitors_foreign"] == 0) & (wide["revenue_foreign_mthb"] > 0),
    "ngoại: visitors>0 nhưng revenue=0": (wide["visitors_foreign"] > 0) & (wide["revenue_foreign_mthb"] == 0),
}
pd.DataFrame({"n_dòng": {k: int(v.sum()) for k, v in checks.items()}})
```

**Nhận xét:** `<bao nhiêu dòng mâu thuẫn, bao nhiêu 0 cô lập>`. Em phân loại: 0 thật khi `<điều kiện>`, thiếu khi `<điều kiện>`.

##### 2.4.4 Quyết định xử lý

```python
w = wide.sort_values(["province_en", "date"]).copy()
g = w.groupby("province_en")["visitors_total"]
w["prev"], w["next"] = g.shift(1), g.shift(-1)

zero_now = w["visitors_total"] == 0
isolated = zero_now & (w["prev"] > 0) & (w["next"] > 0)    # 0 kẹp giữa hai tháng có số liệu
print("visitors_total = 0:", zero_now.sum(), "| trong đó cô lập:", isolated.sum())
print(w[zero_now].groupby(w["date"].dt.year).size())
```

### 2.5 Relationships
Các chỉ số đều tỷ lệ với quy mô tỉnh, nên tương quan trên toàn bảng bị chi phối bởi khác biệt giữa các tỉnh. Em tính tương quan ở ba mức: toàn bảng, trong cùng tỉnh, và thay đổi theo tháng.

##### 2.5.1 Tương quan

```python
wide["year"]  = wide["date"].dt.year
wide["month"] = wide["date"].dt.month

# log cho số lượng và doanh thu, giữ nguyên occupancy (đã nằm trong 0-100)
log_cols = count_cols + rev_cols
feat = wide[num_cols].copy()
feat[log_cols] = np.log1p(feat[log_cols])
```

```python
corr_all = feat.corr(method="spearman")
```

```python
within = feat - feat.groupby(wide["province_en"]).transform("mean")
corr_within = within.corr(method="spearman")
```

```python
chg = feat.groupby(wide["province_en"]).diff()
corr_chg = chg.corr(method="spearman")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, c, t in zip(axes, [corr_all, corr_within, corr_chg],
                    ["Toàn bảng", "Trong cùng tỉnh", "Thay đổi theo tháng"]):
    im = ax.imshow(c.values, vmin=-1, vmax=1, cmap="coolwarm")
    ax.set_xticks(range(len(c))); ax.set_xticklabels(c.columns, rotation=60, ha="right", fontsize=7)
    ax.set_yticks(range(len(c))); ax.set_yticklabels(c.columns, fontsize=7)
    ax.set_title(f"Spearman: {t}")
    for i in range(len(c)):
        for j in range(len(c)):
            ax.text(j, i, f"{c.values[i, j]:.2f}", ha="center", va="center", fontsize=6)
fig.colorbar(im, ax=axes, shrink=0.8, label="Hệ số Spearman")
plt.show()
```

**Nhận xét:** `<các cặp cao ở toàn bảng; cặp nào giảm mạnh khi bỏ quy mô; cặp nào vẫn cao ở cả ba mức>`. Cặp `visitors_total` với `visitors_thai/foreign` tương quan cơ học vì tổng bằng đúng Thái + ngoại (mục 2.2.4).

##### 2.5.2 Biến phái sinh

```python
den = wide["visitors_total"].where(wide["visitors_total"] > 0)
wide["rev_per_visitor_thb"] = wide["revenue_total_mthb"] * 1e6 / den
wide["foreign_share_visitors"] = wide["visitors_foreign"] / den
wide["foreign_share_revenue"] = (wide["revenue_foreign_mthb"] /
                                 wide["revenue_total_mthb"].where(wide["revenue_total_mthb"] > 0))
wide["guests_to_visitors"] = wide["guests"] / den

derived = ["rev_per_visitor_thb", "foreign_share_visitors",
           "foreign_share_revenue", "guests_to_visitors"]
print(wide[derived].describe().round(3))
print(wide[derived].isna().sum())
```

```python
thai_rpv = wide["revenue_thai_mthb"] * 1e6 / wide["visitors_thai"].where(wide["visitors_thai"] > 0)
for_rpv  = wide["revenue_foreign_mthb"] * 1e6 / wide["visitors_foreign"].where(wide["visitors_foreign"] > 0)
print(pd.DataFrame({"thai": thai_rpv.describe(), "foreign": for_rpv.describe()}).round(0))
```
